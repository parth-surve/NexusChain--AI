```python
plt.figure(figsize=(8,5))
sns.countplot(
    x='Shipping Mode',
    hue='Late_delivery_risk',
    data=df
)
plt.title('Shippinng Mode vs Late Delivery Risk')
plt.xticks(rotation=45)
plt.show()
```

Standard class has more count of delivery and more on time dekivery

1st n 2nd  class has more late delivery on a greater scale 

same day delivery has very less count and  data is balanced while on time delivery is leading 

```python
plt.figure(figsize=(8,5))

sns.countplot(
    x="Customer Segment",
    hue="Late_delivery_risk",
    data=df
)

plt.title("Customer Segment vs Late Delivery")
plt.xticks(rotation=15)

plt.show()
```

count - consumer>Corporate> home office 

late delivery dominates 

```python
plt.figure(figsize=(10,5))

sns.countplot(
    y="Market",
    hue="Late_delivery_risk",
    data=df
)

plt.title("Market vs Late Delivery")

plt.show()
```

LATAM and Europe are the biggest markets followed by> Pacific Asia > USCA > Africa

Late deliveries still DOMINATE



```python
plt.figure(figsize=(10,8))

sns.countplot(
    y="Category Name",
    order=df["Category Name"].value_counts().index,
    data=df
)

plt.title("Top Product Categories")

plt.show()
```

```python
plt.figure(figsize=(12,8))

sns.countplot(
    y="Order Region",
    order=df["Order Region"].value_counts().index,
    data=df
)

plt.title("Orders by Region")

plt.show()
```

Central America and Western Europe have the largest counts
